# 2026 Global Job Market Analysis - Step 1: Data Inspection Notebook
This notebook loads the 2026 all-industry job dataset, performs schema validation, checks null values, evaluates duplicates, and determines column utility for Power BI.

```python
import pandas as pd
import numpy as np
from collections import Counter

# Load raw 2026 dataset
df_raw = pd.read_csv('../data/raw/jobs_2026_raw.csv')
print(f"Raw dataset dimensions: {df_raw.shape[0]:,} rows by {df_raw.shape[1]} columns")
df_raw.head(3)
```

## 1. Schema & Data Types Summary

```python
inspection_df = pd.DataFrame({
    'Data_Type': df_raw.dtypes,
    'Null_Count': df_raw.isnull().sum(),
    'Fill_Rate_%': ((len(df_raw) - df_raw.isnull().sum()) / len(df_raw) * 100).round(2),
    'Unique_Values': df_raw.nunique()
})
inspection_df
```

## 2. Duplicate Audit

```python
exact_dupes = df_raw.duplicated().sum()
id_dupes = df_raw.duplicated(subset=['job_id']).sum()
print(f"Exact Duplicate Rows: {exact_dupes}")
print(f"Duplicate Job IDs: {id_dupes}")
```

## 3. High-Cardinality & Categorical Overview

```python
print(f"Unique Companies: {df_raw['company_name'].nunique():,}")
print(f"Unique Countries: {df_raw['country'].nunique():,}")
print(f"Date Range: {df_raw['date_posted'].min()} to {df_raw['date_posted'].max()}")
print(f"Workplace Types: {df_raw['workplace_type'].value_counts(dropna=False).to_dict()}")
```
